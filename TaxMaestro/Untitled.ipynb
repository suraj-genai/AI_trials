```python
import fitz  # PyMuPDF
import re
import json

def extract_text_by_page(pdf_path):
    doc = fitz.open(pdf_path)
    pages_text = []
    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        text = page.get_text("text")
        pages_text.append(text)
    return pages_text

def identify_sections(text):
    """
    Identify section headings and split text accordingly.
    Section headings patterns:
    - Chapter X
    - Numeric headings like 1., 1.1, 2.3.4 etc.
    """
    # Regex pattern for section headings
    # Matches lines starting with "Chapter X" or numeric section headings like "1.", "1.1", "2.3.4"
    pattern = re.compile(r'^(Chapter\s+\d+|(\d+(\.\d+)*))\s+([A-Z].*)', re.MULTILINE)

    # Find all matches with their start positions
    matches = list(pattern.finditer(text))

    sections = []
    for i, match in enumerate(matches):
        start = match.start()
        title = match.group(0).strip()
        end = matches[i+1].start() if i+1 < len(matches) else len(text)
        content = text[start:end].strip()
        sections.append({
            "section_title": title,
            "content": content
        })
    return sections

def main(pdf_path, output_json_path):
    pages_text = extract_text_by_page(pdf_path)
    full_text = "\n".join(pages_text)

    sections = identify_sections(full_text)

    with open(output_json_path, "w", encoding="utf-8") as f:
        json.dump(sections, f, ensure_ascii=False, indent=2)

    print(f"Extracted {len(sections)} sections and saved to {output_json_path}")

if __name__ == "__main__":
    pdf_path = r"D:\Suraj\AI_Trials\Trials\pythonProject\Tax Maestro\income_tax_playbook_complete.pdf"
    output_path = r"redacted\output.json"
    main(pdf_path, output_path)
#     import sys
#     if len(sys.argv) != 3:
#         print("Usage: python chunk_pdf.py <input_pdf_path> <output_json_path>")
#     else:
#         main(sys.argv[1], sys.argv[2])
```
